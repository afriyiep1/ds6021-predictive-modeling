```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

from sklearn.linear_model import LinearRegression
```

```python
planets = pd.read_csv("PlanetsData.csv")

planets
```

```python
px.scatter(
    planets,
    x="distance",
    y="revolution",
    trendline="ols",
    hover_data=["planet"] if "planet" in planets.columns else None,
    title="Revolution vs. Distance"
)
```

```python
y=planets["revolution"]
X=planets[["distance"]]

model_linear = LinearRegression()

model_linear.fit(X, y)

print("Intercept:", model_linear.intercept_)
print("Slope:", model_linear.coef_[0])
print("R²:", model_linear.score(X, y))
```

```python
planets["predicted_linear"] = model_linear.predict(X)

plot_data = planets.sort_values("distance")

fig = px.scatter(
    planets,
    x="distance",
    y="revolution",
    hover_data=["planet"] if "planet" in planets.columns else None,
    title="Revolution vs. Distance: Linear Model"
)

fig.add_trace(
    go.Scatter(
        x=plot_data["distance"],
        y=plot_data["predicted_linear"],
        mode="lines",
        name="Linear Model"
    )
)

fig.update_layout(
    xaxis_title="Distance",
    yaxis_title="Revolution"
)

fig.show()
```

```python
planets["residual_linear"] = (
    planets["revolution"] - planets["predicted_linear"]
)

fig = px.scatter(
    planets,
    x="predicted_linear",
    y="residual_linear",
    title="Residual Plot: Linear Model",
    labels={
        "predicted_linear": "Predicted Revolution",
        "residual_linear": "Residual"
    }
)

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.show()
```

```python
planets["log_distance"] = np.log(planets["distance"])
planets["log_revolution"] = np.log(planets["revolution"])

planets[[
    "distance",
    "revolution",
    "log_distance",
    "log_revolution"
]]
```

```python
X_log = planets[["log_distance"]]
y_log = planets["log_revolution"]

model_log = LinearRegression()

model_log.fit(X_log, y_log)

print("Intercept:", model_log.intercept_)
print("Slope:", model_log.coef_[0])
print("R²:", model_log.score(X_log, y_log))
```

```python
planets["predicted_log"] = model_log.predict(X_log)

plot_data_log = planets.sort_values("log_distance")

fig = px.scatter(
    planets,
    x="log_distance",
    y="log_revolution",
    hover_data=["planet"] if "planet" in planets.columns else None,
    title="Log(Revolution) vs. Log(Distance): Linear Model"
)

fig.add_trace(
    go.Scatter(
        x=plot_data_log["log_distance"],
        y=plot_data_log["predicted_log"],
        mode="lines",
        name="Log-Log Model"
    )
)

fig.update_layout(
    xaxis_title="log(Distance)",
    yaxis_title="log(Revolution)"
)

fig.show()
```

```python
planets["residual_log"] = (
    planets["log_revolution"] - planets["predicted_log"]
)

fig = px.scatter(
    planets,
    x="predicted_log",
    y="residual_log",
    title="Residual Plot: Log-Log Model",
    labels={
        "predicted_log": "Predicted log(Revolution)",
        "residual_log": "Residual"
    }
)

fig.add_hline(
    y=0,
    line_dash="dash"
)

fig.show()
```
