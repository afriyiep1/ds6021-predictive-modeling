```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
```

```python
insurance = pd.read_csv("insurance.csv")

insurance.head()
```

```python
fig = px.scatter(
    insurance,
    x="bmi",
    y="charges",
    color="smoker",
    trendline="ols",
    title="BMI and Insurance Charges by Smoking Status"
)

fig.show()
```

```python
insurance["smoker_yes"] = (
    insurance["smoker"] == "yes"
).astype(int)

insurance[["smoker", "smoker_yes"]].head(10)
```

```python
insurance["bmi_smoker"] = (
    insurance["bmi"] * insurance["smoker_yes"]
)

insurance[
    ["bmi", "smoker", "smoker_yes", "bmi_smoker"]
].head(10)
```

```python
y=insurance["charges"]
X_interaction = insurance[["bmi", "smoker_yes","bmi_smoker"]]

interaction_model = LinearRegression()
interaction_model.fit(X_interaction, y)
```

```python
interaction_coefs = pd.DataFrame({
    "feature": [
        "intercept",
        "bmi",
        "smoker_yes",
        "bmi × smoker"
    ],
    "coefficient": [
        interaction_model.intercept_,
        *interaction_model.coef_
    ]
})

interaction_coefs
```

```python
b0 = interaction_model.intercept_
b1, b2, b3 = interaction_model.coef_

print(
    f"charges_hat = {b0:.2f}"
    f" + {b1:.2f}(BMI)"
    f" + {b2:.2f}(smoker)"
    f" + {b3:.2f}(BMI × smoker)"
)
```
