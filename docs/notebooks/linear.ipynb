```python
import pandas as pd
import numpy as np
import plotly.express as px

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
planets = pd.read_csv("/content/PlanetsData.csv")

planets
```

```python
fig = px.scatter(
    planets,
    x="distance",
    y="revolution",
    text="planet",
    title="Distance from the Sun vs. Revolution Period",
    labels={
        "distance": "Distance from the Sun",
        "revolution": "Revolution Period"
    }
)

fig.update_traces(
    textposition="top center"
)

fig.show()
```

```python
planet_model = LinearRegression()

planet_model.fit(
    planets[["distance"]],
    planets["revolution"]
)

print(
    "Intercept:",
    planet_model.intercept_
)

print(
    "Slope:",
    planet_model.coef_[0]
)
```

```python
planets["predicted_revolution"] = planet_model.predict(
    planets[["distance"]]
)

planets = planets.sort_values("distance")
planets
```

```python
fig = px.scatter(
    planets,
    x="distance",
    y="revolution",
    hover_name="planet",
    title="Planet Distance vs. Revolution"
)

fig.add_scatter(
    x=planets["distance"],
    y=planets["predicted_revolution"],
    mode="lines",
    name="Linear Model"
)

fig.update_layout(
    xaxis_title="Distance",
    yaxis_title="Revolution"
)

fig.show()
```

```python
new_planet_distance = 2200
```

```python
new_planet_revolution = planet_model.predict(
    [[new_planet_distance]]
)[0]

print(
    "Predicted revolution:",
    new_planet_revolution
)
```

```python
fig.add_scatter(
    x=[new_planet_distance],
    y=[new_planet_revolution],
    mode="markers",
    name="New Planet",
    marker=dict(
        size=14,
        color="red"
    )
)

fig.show()
```
