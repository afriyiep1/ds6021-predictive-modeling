```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression
```

```python
planets = pd.read_csv("/content/PlanetsData.csv")

planets
```

```python
x = planets["distance"]
y = planets["revolution"]

n = len(y)
```

```python
planet_model = LinearRegression()

planet_model.fit(
    planets[["distance"]],
    planets["revolution"]
)

ols_b0 = planet_model.intercept_
ols_b1 = planet_model.coef_[0]

print("OLS Intercept:", ols_b0)
print("OLS Slope:", ols_b1)
```

```python
learning_rate=1e-8
```

```python
b0 = 0.0
b1 = 0.0

iterations = 10000


checkpoints = [
    0,
    1,
    5,
    10,
    25,
    50,
    100,
    250,
    500,
    1000,
    2500,
    5000,
    7500,
    10000
]

history = {
    0: {
        "b0": b0,
        "b1": b1
    }
}

for i in range(1, iterations + 1):

    y_hat = b0 + b1 * x


    residual = y - y_hat


    grad_b0 = -(2 / n) * np.sum(
        residual
    )

    grad_b1 = -(2 / n) * np.sum(
        residual * x
    )


    b0 = b0 - learning_rate * grad_b0

    b1 = b1 - learning_rate * grad_b1

    if i in checkpoints:

        history[i] = {
            "b0": b0,
            "b1": b1
        }
```

```python
gd_b0 = b0
gd_b1 = b1

print("OLS")
print("Intercept:", ols_b0)
print("Slope:    ", ols_b1)

print()

print("Gradient Descent")
print("Intercept:", gd_b0)
print("Slope:    ", gd_b1)
```

```python
results = []

for iteration, values in history.items():

    results.append({

        "iteration": iteration,

        "intercept": values["b0"],

        "slope": values["b1"]
    })

results = pd.DataFrame(results)

results
```

```python
x_line = np.linspace(
    x.min(),
    x.max(),
    300
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=x,
        y=y,
        mode="markers",
        name="Observed planets",
        marker=dict(
            size=11,
            color="black"
        )
    )
)
```

```python
for iteration, values in history.items():

    if iteration == iterations:
        continue

    b0_i = values["b0"]
    b1_i = values["b1"]

    y_line = (
        b0_i
        +
        b1_i * x_line
    )

    fig.add_trace(
        go.Scatter(
            x=x_line,
            y=y_line,
            mode="lines",
            name=f"GD iteration {iteration}",
            line=dict(
                color="lightgray",
                width=1
            ),
            opacity=0.65
        )
    )
```

```python
gd_y_line = (
    gd_b0
    +
    gd_b1 * x_line
)

fig.add_trace(
    go.Scatter(
        x=x_line,
        y=gd_y_line,
        mode="lines",
        name=f"Gradient Descent: final ({iterations} iterations)",
        line=dict(
            color="red",
            width=5
        )
    )
)
```

```python
ols_y_line = (
    ols_b0
    +
    ols_b1 * x_line
)

fig.add_trace(
    go.Scatter(
        x=x_line,
        y=ols_y_line,
        mode="lines",
        name="OLS solution",
        line=dict(
            color="blue",
            width=4,
            dash="dash"
        )
    )
)
```

```python
fig.update_layout(

    title=(
        "Gradient Descent Converging "
        "to the OLS Regression Line"
    ),

    xaxis_title="Distance",

    yaxis_title="Revolution",

    template="plotly_white",

    width=950,

    height=650
)

fig.show()
```
